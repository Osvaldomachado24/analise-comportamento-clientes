```python
# Preenche valores nulos de 'Review Rating' com a mediana da própria categoria,
# preservando o contexto do grupo e evitando distorções causadas por valores extremos
df['Review Rating']=df.groupby('Category')['Review Rating'].transform(lambda x:x.fillna(x.median()))
```

```python
# Converte nome das colunas para minuscula
df.columns = df.columns.str.lower()
```

```python
# Substitui os espaços por underline
df.columns=df.columns.str.replace(' ','_')
```

```python
df.head(1)
```

```python
# Dicionário de tradução das colunas
colunas_pt = {
    'customer_id': 'id_cliente',
    'age': 'idade',
    'gender': 'genero',
    'item_purchased': 'item_comprado',
    'category': 'categoria',
    'purchase_amount_(usd)': 'valor_compra_usd',
    'location': 'localizacao',
    'size': 'tamanho',
    'color': 'cor',
    'season': 'estacao',
    'review_rating': 'avaliacao',
    'subscription_status': 'status_assinatura',
    'shipping_type': 'tipo_envio',
    'discount_applied': 'desconto_aplicado',
    'promo_code_used': 'codigo_promocional_usado',
    'previous_purchases': 'compras_anteriores',
    'payment_method': 'metodo_pagamento',
    'frequency_of_purchases': 'frequencia_compras'
}
```

```python
# Renomear colunas
df.rename(columns=colunas_pt,inplace=True)
```

```python
df.head()
```

```python
df.columns
```

```python
# Criar coluna de faixa etária baseada em quantis
rotulos = ['Jovem Adulto', 'Adulto', 'Meia-idade', 'Idoso']

df['faixa_etaria'] = pd.qcut(
    df['idade'],
    q=4,
    labels=rotulos
)
```

```python
df[['idade','faixa_etaria']].head(10)
```

```python
frequency_mapping = { 
    'Fortnightly': 14, 
    'Weekly': 7, 
    'Monthly': 30, 
    'Quarterly': 90,
    'Bi-Weekly': 14, 
    'Annually': 365, 
    'Every 3 Months': 90
 }
df['frequencia_compras_dias'] = df['frequencia_compras'].map(frequency_mapping)
```

```python
df[['frequencia_compras_dias','frequencia_compras']].head(10)
```

```python
# Remove a coluna Redudante 
df=df.drop('codigo_promocional_usado', axis=1)
```
